```python
import json
import pandas as pd
```

```python
tachiyomi = 'static/EditedBackup.json'

with open(tachiyomi, 'r', encoding='utf-8') as f:
    data = json.load(f)
```

```python
manga_list = []
for manga in data['backupManga']:
    manga_title = manga.get('title')
    manga_desc = manga.get('description')
    manga_thumbnail = manga.get('thumbnailUrl')
    manga_genre = manga.get('genre')
    manga_chapter = manga.get('chapters')

    for chapter in manga_chapter:
        #print(chapter)
        manga_chapter_latest = chapter.get('name')
        #print(manga_chapter_latest)
        break

    

    manga_list.append({
        'title': manga_title,
        'description': manga_desc,
        'thumbnail': manga_thumbnail,
        'genre': manga_genre,
        'latest_chapter': manga_chapter_latest
    })

manga_df = pd.DataFrame(manga_list)
print(manga_df.head())
```

```python
import json

# Save manga_list to a JSON file
filepath = "static/manga_list.json"
with open(filepath, 'w', encoding='utf-8') as f:
    json.dump(manga_list, f, ensure_ascii=False, indent=2)

print("Manga list saved to manga_list.json")
```
